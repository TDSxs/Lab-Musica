## Carga y Exploración Inicial de Datos (3.1)

Importamos las librerías necesarias de Pytho:(`pandas`, `numpy`, `matplotlib`, `seaborn`) y cargamos los dos datasets seleccionados de Kaggle


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carga de los archivos con los nombres exactos que tenés en Colab
df_features = pd.read_csv('spotify-tracks-dataset-detailed.csv')
df_global = pd.read_csv('spotify_data clean.csv')

print("=== 1. PRIMERAS FILAS: AUDIO FEATURES ===")
display(df_features.head())

print("\n=== 2. ÚLTIMAS FILAS: AUDIO FEATURES ===")
display(df_features.tail())

print("\n=== 3. PRIMERAS FILAS: GLOBAL DATASET ===")
display(df_global.head())

print("\n=== 4. ÚLTIMAS FILAS: GLOBAL DATASET ===")
display(df_global.tail())
```

#### Análisis inicial (lo que vemos a simple vista)

Mirando lo que nos devolvió el código, notamos varios detalles:

1. **Estructura general:**
   * La tabla `Audio Features` tiene 114.000 filas. Vemos columnas como `track_id`, `artists`, `album_name`, `track_name`, `popularity` y `duration_ms`.
   * La tabla `Global` nos suma datos sobre la popularidad de las canciones, géneros y los seguidores de los artistas.

2. **Cosas raras o incómodas que encontramos:**
   * **La duración está en milisegundos (`duration_ms`):** Es muy poco intuitivo ver números como `230666` o `149610`. Para saber cuánto dura un tema hay que pasarlo a minutos, lo cual es confuso para el análisis.
   * **Temas duplicados:** Hay canciones que aparecen más de una vez porque están en álbumes comunes, sencillos o "Lo mejor de..." (Best of)
   * **Nombres de variables y códigos raros:** Columnas como `track_id` tienen códigos largos (ej. `5SuOikwiRyPMVoIQDJUgSV`) que no nos sirven para leer a simple vista.

Solución y limpieza de datos en Python

Para corregir estos detalles y hacer que los datos sean legibles para nuestro análisis, aplicamos las siguientes transformaciones con Pandas:
1. Creamos una columna nueva (`duration_min`) dividiendo los milisegundos entre 60.000 para tener la duración en minutos reales.
2. Limpiamos los temas duplicados.

*Nota de transparencia:* Se consultó a la IA para la sintaxis de la limpieza de datos (conversión de milisegundos a minutos y eliminación de duplicados con Pandas)

*Prompt utilizado:* "¿Cómo creo una columna en minutos dividiendo duration_ms por 60000 y cómo elimino filas duplicadas en un DataFrame de Pandas?"

```python
# ==========================================
# TRANSFORMACIONES Y CORRECCIONES EN PYTHON
# ==========================================

# 1. Solución a los milisegundos: Convertir duration_ms a minutos (dividiendo por 60.000)
df_features['duration_min'] = df_features['duration_ms'] / 60000

# Redondeamos a 2 decimales para que sea super fácil de leer
df_features['duration_min'] = df_features['duration_min'].round(2)

# 2. Solución a los duplicados: Eliminar canciones repetidas basándonos en el track_id
df_features_limpio = df_features.drop_duplicates(subset=['track_id']).copy()

# 3. Mostramos las columnas corregidas (Nombre del tema, Artista, Popularidad y la nueva Duración en Minutos)
print("=== TABLA CORREGIDA CON LA DURACIÓN EN MINUTOS Y SIN DUPLICADOS ===")
display(df_features_limpio[['track_name', 'artists', 'popularity', 'duration_ms', 'duration_min']].head(10))
```

## Transformaciones y Resumen Estadístico (3.3)

Para poder responder a nuestra pregunta sobre 'la clave del éxito musical', realizamos las siguientes dos transformaciones en los datos:

1. **Creación de la variable `duracion_categoria`:**

Agrupamos la duración en minutos en cuatro rangos acotados (corta, ideal, estándar, larga) para evaluar si existe una duración óptima que maximice la popularidad.
Duraciones que estimamos: Corta (<2:30 min), Ideal (2:30-3:30 min), Estándar (3:30-4:30 min) y Larga (>4:30 min).

2. **Creación de la variable `es_hit` / Nivel de Popularidad:**
 Clasificamos los temas según su nivel de popularidad en 'Hit / Alta Popularidad' (si su popularidad es mayor a 70) o 'Baja/Media Popularidad', lo que nos permite comparar las características de las canciones más exitosas.

A continuación, aplicamos la función `.groupby()` para obtener resúmenes estadísticos agregados según estas categorías.

*Nota*:  Se consultó a la IA para la sintaxis del método `pd.cut()` en la categorización de datos y el agrupamiento múltiple con `.groupby()`.
*Prompt utilizado:* "¿Cómo creo categorías de rangos de minutos usando pd.cut en Pandas y cómo saco el promedio de popularidad por categoría usando groupby?"

```python
# =========================================================
# 3.3.a) TRANSFORMACIONES DE DATOS
# =========================================================

# Transformación 1: Categorizar la duración en minutos
limites = [0, 2.5, 3.5, 4.5, float('inf')]
etiquetas = ['Corta (<2.5 min)', 'Ideal (2.5-3.5 min)', 'Estándar (3.5-4.5 min)', 'Larga (>4.5 min)']

df_features_limpio['duracion_categoria'] = pd.cut(df_features_limpio['duration_min'], bins=limites, labels=etiquetas)

# Transformación 2: Categorizar el nivel de popularidad
df_features_limpio['nivel_popularidad'] = pd.cut(
    df_features_limpio['popularity'],
    bins=[-1, 40, 70, 100],
    labels=['Baja', 'Media', 'Alta (Hit)']
)

print("=== MUESTRA DE LAS NUEVAS COLUMNAS ===")
display(df_features_limpio[['track_name', 'duration_min', 'duracion_categoria', 'popularity', 'nivel_popularidad']].head())

# =========================================================
# 3.3.b) RESÚMENES ESTADÍSTICOS AGRUPADOS (.groupby)
# =========================================================

print("\n=== RESUMEN 1: Popularidad promedio por Rango de Duración ===")
resumen_duracion = df_features_limpio.groupby('duracion_categoria', observed=False)['popularity'].agg(['mean', 'median', 'count']).round(2)
display(resumen_duracion)

print("\n=== RESUMEN 2: Características de Audio (Danceability y Energy) por Nivel de Popularidad ===")
resumen_sonido = df_features_limpio.groupby('nivel_popularidad', observed=False)[['danceability', 'energy', 'valence']].mean().round(3)
display(resumen_sonido)
```

#### Gráfico 3: Relación entre las características de audio y la popularidad

Analizamos cómo cambia la popularidad promedio a medida que aumentan la bailabilidad, la energía y la positividad (valence) de las canciones. Como son casi 90.000 canciones, un gráfico de dispersión punto por punto sería ilegible. Por eso dividimos cada variable en 10 rangos (de 0 a 1) y graficamos la popularidad promedio de cada rango como líneas superpuestas. Excluimos los rangos con menos de 100 canciones, porque su promedio no es confiable.

*Nota:* Se consultó a la IA (Claude) para la construcción del gráfico de líneas superpuestas con rangos.
*Prompt utilizado:* "Armame el código de un gráfico de líneas que muestre la popularidad promedio según rangos de danceability, energy y valence, con el mismo estilo que mis gráficos anteriores en seaborn."

```python
# =========================================================
# GRÁFICO 3: Popularidad promedio según características de audio
# =========================================================

# Dividimos el rango 0-1 en 10 tramos iguales (0-0.1, 0.1-0.2, ..., 0.9-1)
bordes = np.linspace(0, 1, 11)
centros = (bordes[:-1] + bordes[1:]) / 2   # punto medio de cada tramo, para el eje X

# Cada característica con su nombre, su color y su marcador
variables = {
    'danceability': ('Bailabilidad', '#FFA800', 'o'),          # círculo
    'energy':       ('Energía', '#f76800', 's'),               # cuadrado
    'valence':      ('Positividad (valence)', '#f71600', '^')  # triángulo
}

sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 5))

# Fondo negro (de la figura y del área del gráfico)
fig.patch.set_facecolor('black')
ax.set_facecolor('black')

for columna, (nombre, color, marcador) in variables.items():
    # Asignamos cada canción a su tramo
    rangos = pd.cut(df_features_limpio[columna], bins=bordes, include_lowest=True)

    # Popularidad promedio y cantidad de canciones por tramo
    resumen = df_features_limpio.groupby(rangos, observed=False)['popularity'].agg(['mean', 'count'])
    resumen['centro'] = centros

    # Descartamos tramos con pocas canciones (promedios poco confiables)
    resumen = resumen[resumen['count'] >= 100]

    ax.plot(resumen['centro'], resumen['mean'], marker=marcador, linewidth=2,
            color=color, label=nombre)

# Título y ejes en blanco
ax.set_title('Popularidad Promedio según Características de Audio',
             fontsize=14, fontweight='bold', color='white')
ax.set_xlabel('Valor de la característica (0 = mínimo, 1 = máximo)', fontsize=12, color='white')
ax.set_ylabel('Popularidad Promedio (0 a 100)', fontsize=12, color='white')
ax.set_xticks(bordes)
ax.tick_params(colors='white')   # números de los ejes en blanco

# Grilla gris oscuro y bordes del gráfico en gris
ax.grid(color='#444444')
for borde in ax.spines.values():
    borde.set_color('#888888')

# Leyenda con fondo negro y letras blancas
leyenda = ax.legend(title='Característica', facecolor='black', edgecolor='#888888', labelcolor='white')
leyenda.get_title().set_color('white')

# Guardar la imagen para el informe (con el fondo negro incluido)
plt.savefig('grafico3_popularidad_vs_audio.png', dpi=300, bbox_inches='tight',
            facecolor=fig.get_facecolor())

plt.show()
```

#### Interpretación del Gráfico 3
Las tres características muestran una relación con forma de "U invertida" con la popularidad: el promedio crece desde los valores bajos, alcanza un máximo en una zona intermedia y vuelve a caer en los valores extremos.

* **Energía:** alcanza la popularidad promedio más alta (≈35,5 puntos) entre 0,4 y 0,8, mientras que las canciones de energía muy baja (0,1-0,2) o máxima (>0,9) rondan los 27-30 puntos.
* **Bailabilidad:** las canciones poco bailables (<0,3) tienen la menor popularidad (≈27 puntos); el máximo se ubica entre 0,5 y 0,8 (≈34 puntos), y las extremadamente bailables vuelven a bajar.
* **Positividad (valence):** su pico aparece antes (entre 0,2 y 0,6), lo que indica que las canciones más populares tienden a un tono emocional neutro o levemente melancólico, y no a los extremos de euforia o tristeza.

En conjunto, el patrón que comparten las canciones más populares es el **equilibrio**: bailables y enérgicas, pero sin valores extremos. Este resultado es coherente con lo observado en la duración, donde también predominaba el rango intermedio. Cabe aclarar que el eje vertical no empieza en cero: la diferencia entre el tramo menos y el más popular es de unos 9 puntos sobre 100, por lo que la relación existe pero es moderada.

### Gráfico 4: Mapa de correlaciones entre variables

Para tener una visión general, calculamos la correlación entre la popularidad y todas las características numéricas de las canciones. La correlación va de -1 a 1: cerca de 1 indica que las dos variables suben juntas; cerca de -1, que cuando una sube la otra baja; cerca de 0, que no hay relación lineal. Excluimos `key` y `time_signature` porque, aunque son números, representan categorías (tonalidad y compás) y su correlación no tiene sentido.

*Nota:* Se consultó a la IA (Claude) para la construcción del mapa de calor de correlaciones.
*Prompt utilizado:* "Armame el código de un heatmap de correlaciones entre popularity y las características de audio, mostrando solo la mitad inferior de la matriz."

```python
# =========================================================
# GRÁFICO 4: Mapa de correlaciones (heatmap)
# =========================================================

columnas = ['popularity', 'duration_min', 'explicit', 'danceability', 'energy',
            'loudness', 'mode', 'speechiness', 'acousticness',
            'instrumentalness', 'liveness', 'valence', 'tempo']

# explicit es True/False: lo pasamos a 1/0 para poder correlacionarlo
datos_corr = df_features_limpio[columnas].copy()
datos_corr['explicit'] = datos_corr['explicit'].astype(int)

correlaciones = datos_corr.corr().round(2)

# Ocultamos la mitad superior (es un espejo de la inferior)
mascara = np.triu(np.ones_like(correlaciones, dtype=bool))

plt.figure(figsize=(11, 8))
sns.heatmap(
    correlaciones,
    mask=mascara,
    annot=True, fmt='.2f',
    cmap='coolwarm', center=0, vmin=-1, vmax=1,
    linewidths=0.5,
    cbar_kws={'label': 'Correlación'}
)

plt.title('Correlación entre la Popularidad y las Características de las Canciones', fontsize=14, fontweight='bold')
plt.xlabel('Variable', fontsize=12)
plt.ylabel('Variable', fontsize=12)

plt.savefig('grafico4_heatmap_correlaciones.png', dpi=300, bbox_inches='tight')
plt.show()

# Ranking: qué variables se relacionan más con la popularidad
print("=== CORRELACIÓN DE CADA VARIABLE CON LA POPULARIDAD (de mayor a menor) ===")
display(correlaciones['popularity'].drop('popularity').sort_values(ascending=False))
```

#### Interpretación del Gráfico 4
Las correlaciones entre la popularidad y las características de las canciones son muy débiles: todas se ubican entre -0,13 y 0,07. La más relevante es `instrumentalness` (-0,13), lo que indica que las canciones instrumentales tienden a ser algo menos populares que las cantadas. Ninguna característica, por sí sola, explica el éxito de una canción.

Este resultado no contradice al Gráfico 3, sino que lo complementa. La correlación solo mide relaciones lineales ("cuanto más, más"), mientras que la energía, la bailabilidad y la positividad mostraron una relación en forma de "U invertida" con la popularidad: crece hasta un punto intermedio y luego disminuye. Al compensarse ambos tramos, la correlación resulta cercana a cero (por ejemplo, 0,01 para la energía), aunque la relación existe.

Por otro lado, se observan correlaciones fuertes entre las propias características: la energía se asocia positivamente con el volumen (0,76) y negativamente con la acústica (-0,73), y la bailabilidad se asocia con la positividad (0,49). Esto indica que varias variables describen aspectos similares del sonido.

En conjunto, los resultados sugieren que el éxito de una canción no depende de una única característica sonora, sino de una combinación equilibrada de ellas y, probablemente, de factores externos que el dataset no captura (popularidad del artista, promoción, presencia en playlists).

### Gráfico 5: ¿Pesa más el artista que la canción? Seguidores vs. popularidad

El análisis anterior mostró que ninguna característica sonora, por sí sola, explica la popularidad. Para evaluar la influencia de factores externos al sonido, usamos el segundo dataset (Spotify Global Music Dataset) y analizamos la relación entre la cantidad de seguidores del artista (`artist_followers`) y la popularidad de la canción (`track_popularity`).

Transformaciones aplicadas:
1. **Eliminación de duplicados y nulos:** quitamos canciones repetidas por `track_id` y filas sin datos de seguidores o popularidad.

Por la misma razón, en el gráfico de dispersión usamos una **escala logarítmica** en el eje de seguidores: cada marca del eje es 10 veces la anterior. Con escala normal, casi todos los puntos quedarían amontonados a la izquierda.

##Creación de Segmentos: Aislamiento de la Fama

Para que los artistas gigantes no distorsionen los resultados y podamos medir qué hace que una canción funcione por sí sola, definimos umbrales claros basados en percentiles:

Filtro de artistas: Excluimos al top 25% de los artistas con más seguidores (quedándonos por debajo del percentil 75). De esta forma, enfocamos el estudio únicamente en artistas emergentes o de tamaño medio.

División de los grupos: Dentro de este grupo de artistas emergentes, separamos las canciones en dos categorías para compararlas:

-Breakout Hits (Éxitos Revelación): Canciones que lograron meterse en el top 25% de popularidad del segmento.

-Canciones Normales (Grupo de control): El 75% restante de las canciones que no alcanzaron esa viralidad.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual
sns.set_theme(style="darkgrid", palette="pastel")
plt.rcParams['figure.figsize'] = (12, 6)

# 1. Cargar y fusionar datos
df_detailed = pd.read_csv("spotify-tracks-dataset-detailed.csv")
df_final = pd.read_csv("track_data_final.csv")

# Cruzamos por track_id para tener características acústicas y seguidores del artista en un solo lugar
df = pd.merge(df_detailed, df_final[['track_id', 'artist_popularity', 'artist_followers']], on='track_id', how='inner')
df.drop_duplicates(subset=['track_id'], inplace=True)

print(f"Dataset consolidado: {df.shape[0]} canciones.")
```

```python
# Definir umbrales
followers_75 = df['artist_followers'].quantile(0.75)
pop_75 = df['popularity'].quantile(0.75)

# Filtrar solo artistas que no son mega-estrellas (aislamos el factor fama)
df_emerging = df[df['artist_followers'] <= followers_75].copy()

# Clasificar las canciones
df_emerging['Segmento'] = np.where(df_emerging['popularity'] >= pop_75, 'Breakout Hit 🔥', 'Normal 🎵')

hits_count = df_emerging[df_emerging['Segmento'] == 'Breakout Hit 🔥'].shape[0]
normal_count = df_emerging[df_emerging['Segmento'] == 'Normal 🎵'].shape[0]

print(f"Artistas no mega-famosos analizados: {df_emerging.shape[0]} canciones")
print(f"- Breakout Hits: {hits_count}")
print(f"- Canciones Normales: {normal_count}")
```

### Conclusión: La Fórmula del Éxito Orgánico

Para meter un hit sin depender de ser un artista famoso, el análisis de datos nos deja 3 claves fundamentales:

1. **Perfil del Sonido:** El tema tiene que ser bailable (*danceability* alta) y mantener una velocidad de **110 a 130 BPM**. Es indispensable que haya voz (los temas instrumentales puros no funcionan acá) y garpan mucho las melodías con un toque más serio o melancólico por sobre la euforia exagerada.
2. **Duración Ideal:** Hay que ir directo al grano. La duración óptima se ubica entre los **2:30 y 3:15 minutos** (150 a 195 segundos). Canciones más largas corren el riesgo de aburrir al oyente y perjudicar la recomendación del algoritmo.
3. **Nicho y Autenticidad:** Es mejor apostar por géneros con nichos bien marcados (Dance, Alt-Rock, Hip-Hop) en lugar de competir en Pop masivo. Además, escribir letras sin censura no frena el alcance del tema.

```python
# =========================================================
# TOP 25 % DE ARTISTAS CON MÁS SEGUIDORES
# =========================================================

# 1. Un registro por artista (el dataset tiene una fila por canción)
#    Se toma el máximo de seguidores porque el mismo artista puede tener valores apenas distintos entre filas
artistas = (df_artistas
            .groupby('artist_name')
            .agg(seguidores=('artist_followers', 'max'),
                 popularidad_artista=('artist_popularity', 'max'),
                 canciones=('track_id', 'count'),
                 popularidad_prom_canciones=('track_popularity', 'mean'))
            .round(1)
            .reset_index())

# 2. Umbral del 25 % superior (percentil 75)
umbral = artistas['seguidores'].quantile(0.75)

# 3. Filtrar y ordenar de mayor a menor
top_25 = (artistas[artistas['seguidores'] >= umbral]
          .sort_values('seguidores', ascending=False)
          .reset_index(drop=True))
top_25.index = top_25.index + 1   # que el ranking empiece en 1

print(f"Artistas en total: {len(artistas)}")
print(f"Umbral del 25 % superior: {umbral:,.0f} seguidores".replace(',', '.'))
print(f"Artistas en el top 25 %: {len(top_25)}")

# 4. Mostrar el listado completo
pd.set_option('display.max_rows', None)
display(top_25)
```
